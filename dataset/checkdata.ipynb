```python
import pandas as pd
df=pd.read_excel('C:/-----SEM4-----/projects/data eng/online_retail_II.xlsx')
df.head()
```

```python
keys= [cols for cols in df.columns if df[cols].is_unique and df[cols].notna().all()]
print("primary key could be: ",keys)
```

```python
for col in df.columns:
    unique_count = df[col].nunique()
    total_count = len(df)
    missing_count = df[col].isna().sum()
    
    print(f"Column: '{col}'")
    print(f"  - Unique values: {unique_count} out of {total_count}")
    print(f"  - Missing values: {missing_count}\n")
```

```python
# Check if the combination of Invoice and StockCode is 100% unique
is_composite_unique = not df.duplicated(subset=['Invoice', 'StockCode']).any()

print("Is Invoice + StockCode a valid primary key?:", is_composite_unique)
```
